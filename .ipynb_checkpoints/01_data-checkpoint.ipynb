- We are trying to predict machine failure, how often do machines fail?
- Don't need UIDs
- Product ID are strs, the start is either L,M, H representing low, medium, high, this is in Type
- We will need to one hot encode L, M, H.
- Convert kelvins to C so its easier to interpret
- We have Type, Air temp, process temp, rotational speed, torque and tool wear which will be used to help to be used for machine failure, and then TWF, HDF, PWF, OSF, RNF are the specific failure types.
- Air temp is outside air temp
- Process temp is the operating temperature of engine
- type is product quality
- rotational speed is how many full turns the crankshaft makes per min
- torque is the twisting force the crankshaft produces
- total wear probably linked to type, may be redundent 

```python
df = df.drop(columns = ['UDI', 'Product ID'])
```

```python
df[['Air temperature [K]', 'Process temperature [K]']] = df[['Air temperature [K]', 'Process temperature [K]']].apply(lambda x: x - 273.15)
```

```python
df = df.rename(columns = {'Air temperature [K]': 'Air_temp_C', 'Process temperature [K]': 'Process_temp_C'})
```

Relabel target and check for imbalance. 

```python
df['Machine failure'].mean() * 100 
```

```python
df = df.rename(columns = {'Machine failure': 'Target'})
```

**Target comments**

- We have an imbalanced dataset. Only 3.4\% are failures (approx 340 failures). Accuracy will be high by just predicting no failure, we would get accuracy of approx 96.6. Need to focus precision and recall instead.
- Precision-recall curve is best choice when dataset is imbalanced. ROC Curve focuses on TPR vs FPR. TPR = TP / (TP + FN) (recall) and FPR = FP / (FP + TN). For imbalanced dataset, with many negatives, FPR will remain low even if we missclassify many results as positive.
- Question for client: "What is more costly, missing a failure, or having a false alert" this may change what our threshold for detection is.
- If cost of false positive (send a team) is C_FP, and cost of FN (miss a fault) is C_FN, then send a crew when

    C_FN * p > C_FP * (1-p)
    
    which gives send crew when p > C_FP / (C_FP + C_FN). This approach assumes probabilities are calibrated. Usually aren't so we may need to fix. 

- Imbalanced dataset means we need to be mindful of train, validation and test splits. A 20% test dataset means only 68 failures, if we use proportional split (this could be more or less if its random). Small variation if model predictions will cause large differences in precision and recall. Using cross validation with 5-10 folds would lower the variance of the performance estimates by averaging the validation set performance across multiple training folds.

- We should do 80/20 stratified train test split. Run a 5 fold stratified CV on train set to tune and select best model based on PR-AUC (average precision). Refit final model on full train and get test set prediction.

- We could bootstrap estimate CI for final test set estimates.

```python
# Drop rows where Target == 0 but RNF == 1 
df = df.drop(df[(df['Target'] == 0) & (df['RNF'] == 1)].index)
```

```python
df = df.drop(columns = fail_flags_label)
```

```python
df.columns = df.columns.str.lower().str.strip().str.replace(' ', '_')
```

```python
df.head()
```

Feature engineering 
- HDF is heat dissipation failure, we are interested in temperature gap rather than air temp or process temp.
- PWF is power failure, too low and machine doesnt cut and too high and we overload machine
- Power = speed x torque x 2pi/60
- OSF is overstrain failure, strain = tool wear x torque could be more informative
- TWF is tool wear failure probably heavily linked to tool wear

```python
df['temp_diff_c'] = df['process_temp_c'] - df['air_temp_c']
```

```python
df = df.drop(columns =['process_temp_c'])
```

```python
import numpy as np
```

```python
df['power'] = df['rotational_speed_[rpm]'] * df['torque_[nm]'] * (2 * np.pi) / 60
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

Plot of temperature difference for failed vs healthy machines. 

```python
sns.histplot(data = df, x = 'temp_diff_c', hue = 'target', 
             stat = 'density', common_norm = False, alpha = 0.5)
plt.show()
```

```python
sns.scatterplot(data=df.sort_values('target'), x='torque_[nm]',
                y='rotational_speed_[rpm]', hue='target', alpha=0.5)
```

Failure sits at two ends, high toque low rpm (machine working hard), low torque high rpm (not enough power to cut properly). 

```python
ax = sns.scatterplot(data=df[df['target'] == 0], x='power', y='tool_wear_[min]',
                     alpha=0.2, s=10, label='No failure')
sns.scatterplot(data=df[df['target'] == 1], x='power', y='tool_wear_[min]',
                alpha=0.9, s=25, label='Failure', ax=ax)
```

```python
ax = sns.scatterplot(data=df[df['target'] == 0], x='torque_[nm]', y='tool_wear_[min]',
                     alpha=0.2, s=10, label='No failure')
sns.scatterplot(data=df[df['target'] == 1], x='torque_[nm]', y='tool_wear_[min]',
                alpha=0.9, s=25, label='Failure', ax=ax)
```

```python
df['strain'] = df['torque_[nm]'] * df['tool_wear_[min]']
```

```python
sns.histplot(data = df, x = 'strain', hue = 'target', stat='density', common_norm=False)
```

Extreme (too low) or (too high) power can be linked to failure. Higher tool wear -> higher power cluster of failures.
